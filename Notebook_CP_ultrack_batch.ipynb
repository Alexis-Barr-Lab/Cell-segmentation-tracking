```python
from pathlib import Path
import subprocess
from tqdm import tqdm
import sys

input_folder = Path("/path/to/image/folder")
movie_files = list(input_folder.glob("*.tif"))

for movie_file in tqdm(movie_files, desc="Processing Movies"):
    result = subprocess.run(
        [sys.executable, "/path/to/script/Process_movie_multichannel.py", str(movie_file)],
        capture_output=True, text=True
    )
    if result.returncode != 0:
        print(f"\nError processing {movie_file.name}:")
        print(result.stderr)
```
